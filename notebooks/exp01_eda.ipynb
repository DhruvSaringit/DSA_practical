# Experiment 01: Exploratory Data Analysis (EDA) on a Real-World Dataset Using Python

**Aim:** To perform exploratory data analysis on the Medical Insurance Cost dataset, inspect structural properties, detect missing values, calculate summary statistics, evaluate bivariate correlations, and derive data-driven insights.

### Dataset Description
**Dataset:** Medical Insurance Cost Dataset (1,338 records, 7 features).
- `age`: Primary beneficiary age (18-64).
- `sex`: Gender (male/female).
- `bmi`: Body Mass Index in kg/m^2 (15.96 - 53.13).
- `children`: Number of dependent children covered (0-5).
- `smoker`: Smoking status (yes/no).
- `region`: Residential zone in the US (northeast, northwest, southeast, southwest).
- `charges`: Individual medical expenditures billed by insurance ($1,121.87 - $63,770.43).

### Theory
Exploratory Data Analysis (EDA) is the foundational phase in data analytics that systematically assesses data structure, missingness, distribution shape, and variable associations prior to statistical modeling. Summary metrics such as mean, median, standard deviation, and interquartile ranges describe central tendency and dispersion. Pearson correlation coefficient r measures pairwise linear dependence within [-1, +1]. Handling strategies must account for missingness patterns (MCAR, MAR, MNAR) and preserve data integrity.

#### 1. Load Dataset and Inspect Structural Attributes

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../datasets/insurance.csv')
print('Dataset Dimensions (rows, cols):', df.shape)
print('\nData Information:')
df.info()
print('\nFirst 5 Records:')
df.head()
```

#### 2. Missing Value and Anomaly Detection

```
print('Missing Values per Column:')
print(df.isnull().sum())
print('\nUnique values per categorical feature:')
for col in ['sex', 'smoker', 'region']:
    print(f'{col}: {df[col].unique()}')
```

#### 3. Statistical Summaries and Group Aggregations

```
print('Numerical Features Summary Statistics:')
print(df.describe().round(2))
print('\nCharges Summary by Smoking Status:')
print(df.groupby('smoker')['charges'].agg(['count', 'mean', 'median', 'std']).round(2))
print('\nCharges Summary by Region:')
print(df.groupby('region')['charges'].agg(['count', 'mean']).round(2))
```

#### 4. Bivariate Correlation Matrix and Heatmap Visualization

```
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr().round(3)
print('Correlation Matrix:\n', corr)

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Insurance Features')
plt.tight_layout()
plt.show()
```

### Observations
1. The Medical Insurance dataset contains 1,338 complete observations across 7 features with 0 missing values.
2. Beneficiaries who smoke incur average medical charges of $32,050.23, compared to $8,434.27 for non-smokers (nearly a 4x differential).
3. Age exhibits a moderate positive correlation with insurance charges (r = +0.299), reflecting increased health risks with age.
4. BMI shows a positive linear correlation (r = +0.198) with charges, which compounds significantly in the smoker cohort.

### Result
Comprehensive EDA revealed that smoking status is the primary explanatory determinant of high medical insurance charges, followed by patient age and BMI. The dataset requires zero missing value imputation, and numerical distributions are consistent with actuarial health metrics.
