# Experiment 02: Visualization of Data Using Charts, Histograms, and Scatter Plots in Python

**Aim:** To visualize geometric morphological attributes of wheat varieties from the Seeds dataset using histograms, scatter plots, box plots, and categorical bar charts in Python.

### Dataset Description
**Dataset:** UCI Seeds Morphology Dataset (210 wheat kernels, 7 geometric continuous measurements, 3 wheat varieties: 1: Kama, 2: Rosa, 3: Canadian).
- `area`: Kernel area (10.59 - 21.18 mm^2).
- `perimeter`: Kernel perimeter (12.41 - 17.25 mm).
- `compactness`: Shape compactness C = 4*pi*A/P^2 (0.808 - 0.918).
- `length_kernel`: Kernel length (4.89 - 6.67 mm).
- `width_kernel`: Kernel width (2.63 - 4.03 mm).
- `asymmetry_coef`: Asymmetry coefficient (0.765 - 8.456).
- `groove_length`: Groove length (4.51 - 6.55 mm).
- `variety`: Botanical variety label (1, 2, 3; 70 instances each).

### Theory
Data visualization leverages graphical perception (position, length, hue, area) to uncover multidimensional structure and data distributions. Histograms estimate probability density functions; box plots display median, quartiles, and IQR-based outliers; scatter plots reveal bivariate clustering; and bar charts display aggregated class statistics.

#### 1. Load Dataset and Assign Feature Headers

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

cols = ['area', 'perimeter', 'compactness', 'length_kernel', 'width_kernel', 'asymmetry_coef', 'groove_length', 'variety']
df = pd.read_csv('../datasets/seeds.csv', header=None, names=cols)
variety_map = {1: 'Kama', 2: 'Rosa', 3: 'Canadian'}
df['variety_name'] = df['variety'].map(variety_map)
print(df.head())
print('\nClass distribution:\n', df['variety_name'].value_counts())
```

#### 2. Histograms with Kernel Density Estimation (KDE)

```
plt.figure(figsize=(12, 8))
for i, col in enumerate(cols[:6], 1):
    plt.subplot(2, 3, i)
    sns.histplot(df[col], kde=True, color='teal')
    plt.title(f'Histogram of {col}')
plt.tight_layout()
plt.show()
```

#### 3. Bivariate Scatter Plot Colored by Wheat Variety

```
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='length_kernel', y='width_kernel', hue='variety_name', style='variety_name', s=70, palette='Set2')
plt.title('Wheat Kernel Length vs Width by Botanical Variety')
plt.xlabel('Length of Kernel (mm)')
plt.ylabel('Width of Kernel (mm)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### 4. Multi-Panel Box Plots and Group Feature Means Bar Chart

```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='variety_name', y='area', palette='Set3', ax=axes[0])
axes[0].set_title('Kernel Area Distribution Across Varieties')
axes[0].set_ylabel('Area (mm^2)')

df.groupby('variety_name')[cols[:4]].mean().plot(kind='bar', ax=axes[1], colormap='viridis')
axes[1].set_title('Mean Morphological Measurements per Variety')
axes[1].set_ylabel('Mean Value (mm / mm^2)')
plt.tight_layout()
plt.show()
```

### Observations
1. The Seeds dataset exhibits balanced 70:70:70 class distributions across Kama, Rosa, and Canadian wheat varieties.
2. Morphological histograms for area and perimeter display distinct multimodality, indicating underlying class separation.
3. The scatter plot of kernel length versus kernel width cleanly separates Rosa (largest kernels: mean length 6.02 mm, width 3.64 mm) from Canadian (smallest: length 5.12 mm, width 2.87 mm).
4. Kama occupies an intermediate dimensional tier with mild overlap with Canadian kernels.

### Result
Visual exploration confirms that geometric attributes provide clear discriminatory boundaries for wheat classification. Scatter plots and box plots effectively distinguish species without requiring non-linear projections.
